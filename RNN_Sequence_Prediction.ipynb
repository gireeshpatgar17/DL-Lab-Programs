```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
from tensorflow.keras.utils import to_categorical

text = "HELLO"

chars = sorted(list(set(text)))

char_to_index = {c:i for i, c in enumerate(chars)}
index_to_char = {i:c for i, c in enumerate(chars)}

print("Character to index mapping:")
print(char_to_index)
print("\nIndex to character mapping:")
print(index_to_char)

X = []
y = []

for i in range(len(text) - 1):
  X.append(char_to_index[text[i]])
  y.append(char_to_index[text[i+1]])

X = np.array(X)
y = np.array(y)

num_classes = len(chars)
X = to_categorical(X, num_classes=num_classes)
y = to_categorical(y, num_classes=num_classes)

X = X.reshape(X.shape[0], 1, X.shape[1])

print("Input Shape:", X.shape)

model = Sequential()

model.add(SimpleRNN(
    units=8,
    input_shape=(1, num_classes),
    activation='relu'
))

model.add(Dense(
    num_classes,
    activation='softmax'
))

model.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.fit(
    X,
    y,
    epochs=500,
    verbose=1
)

def predict_next_char(char):
  x = char_to_index[char]
  x = to_categorical([x], num_classes=num_classes)
  x = x.reshape(1, 1, num_classes)
  prediction = model.predict(x, verbose = 0)
  index = np.argmax(prediction)
  return index_to_char[index]

print("\n Testing Preiction:")

test_char = "H"
next_char = predict_next_char(test_char)
print("Input:", test_char)
print("Predicted Next Character:", next_char)


print("\nGenerated Text:")
char = "H"
generated_text = char
for i in range(4):
 next_char = predict_next_char(char)
 generated_text += next_char
 char = next_char
print(generated_text)
```
