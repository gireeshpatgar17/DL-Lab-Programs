```python
import numpy as np

def relu(x):
    return 1 / (1 + np.exp(-x))

def relu_derivative(x):
    return x * (1 - x)

X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

np.random.seed(0)

W1 = np.random.uniform(-1, 1, (2, 4))
W2 = np.random.uniform(-1, 1, (4, 1))

lr = 0.1

for epoch in range(20000):

    hidden_layer = relu(X @ W1)
    output_layer = relu(hidden_layer @ W2)

    error = y - output_layer

    d_output = error * relu(output_layer)
    d_hidden = (d_output @ W2.T) * relu_derivative(hidden_layer)

    W2 += hidden_layer.T @ d_output * lr
    W1 += X.T @ d_hidden * lr

    if epoch % 10000 == 0:
        print(f"Epoch {epoch}, Loss: {np.mean(np.abs(error)):.4f}")

print("\nFinal Output after Training:")
print(output_layer.round(3))


```
